```python
s1_lookup = s1.set_index('entity_id')
s2_lookup = s2.set_index('entity_id')
s3_lookup = s3.set_index('entity_id')


def show_matches(s1_id):

    print('=' * 100)

    s1_row = s1_lookup.loc[s1_id]

    # S1 record
    print("S1")
    print("ID      :", s1_id)
    print("NAME    :", s1_row["business_name"])
    print("ADDRESS :", s1_row["business_address"])
    print("COUNTRY :", s1_row["country"])

    # Ground truth
    truth_row = truth[truth["source1_entity_id"] == s1_id]

    matched_ids = truth_row.iloc[0]["matched_entity_ids"].split(",")

    print("\nTRUE MATCHES:")

    for entity_id in matched_ids:

        entity_id = entity_id.strip()

        if entity_id.startswith("S2-"):
            row = s2_lookup.loc[entity_id]

        elif entity_id.startswith("S3-"):
            row = s3_lookup.loc[entity_id]

        print("\n", entity_id)
        print("NAME    :", row["business_name"])
        print("ADDRESS :", row["business_address"])
        print("COUNTRY :", row["country"])
```

```python
sample_truth = truth.sample(10000, random_state=42)
```

```python
## creating true pairs
pairs = []

for _, row in sample_truth.iterrows():

    s1_id = row["source1_entity_id"]
    matched_ids = str(row["matched_entity_ids"]).split(",")

    s1_row = s1_lookup.loc[s1_id]

    for entity_id in matched_ids:

        entity_id = entity_id.strip()

        if entity_id.startswith("S2-"):
            matched_row = s2_lookup.loc[entity_id]

        elif entity_id.startswith("S3-"):
            matched_row = s3_lookup.loc[entity_id]

        else:
            continue

        pairs.append({
            "s1_name": s1_row["business_name"],
            "matched_name": matched_row["business_name"],
            "s1_address": s1_row["business_address"],
            "matched_address": matched_row["business_address"],
        })

sample_pairs = pd.DataFrame(pairs)
```

```python
## robust normalization(robust normalized name + core name)
import re
import unicodedata

def normalize_name(name):
    if pd.isna(name):
        return ""

    name = str(name).lower()

    # Normalize accented characters
    name = unicodedata.normalize("NFKD", name)
    name = "".join(
        ch for ch in name
        if not unicodedata.combining(ch)
    )

    # Remove URLs
    name = re.sub(r"https?://\S+|www\.\S+", " ", name)

    # Replace punctuation with spaces
    name = re.sub(r"[^a-z0-9\s]", " ", name)

    # Remove extra spaces
    name = re.sub(r"\s+", " ", name).strip()

    return name
```

```python
test_names = [
    "FQ Social, Inc",
    "FQ Social, Inc.",
    "FQ Social, Ínc",
    "Arcxylo DBA FQ Social, Inc"
]

for name in test_names:
    print(name, " -> ", normalize_name(name))
```

```python
## extracting suffixes
SUFFIXES = [
    "private limited",
    "private ltd",
    "pvt limited",
    "pvt ltd",
    "limited",
    "ltd",
    "corporation",
    "corp",
    "incorporated",
    "inc",
    "company",
    "co",
    "llc",
    "llp"
]


def extract_suffix(name):

    name = normalize_name(name)

    for suffix in sorted(SUFFIXES, key=len, reverse=True):

        if name.endswith(" " + suffix):
            return suffix

    return ""
```

```python
for name in test_names:
    print(name, " -> ", extract_suffix(name))
```

```python
## create the core names
def extract_core_name(name):

    name = normalize_name(name)

    suffix = extract_suffix(name)

    if suffix:
        name = name[:-len(suffix)].strip()

    return name
```

```python
sample_pairs["s1_norm"] = sample_pairs["s1_name"].apply(normalize_name)
sample_pairs["matched_norm"] = sample_pairs["matched_name"].apply(normalize_name)

sample_pairs["s1_suffix"] = sample_pairs["s1_name"].apply(extract_suffix)
sample_pairs["matched_suffix"] = sample_pairs["matched_name"].apply(extract_suffix)

sample_pairs["s1_core"] = sample_pairs["s1_name"].apply(extract_core_name)
sample_pairs["matched_core"] = sample_pairs["matched_name"].apply(extract_core_name)
```

```python
## get better names

def remove_numbers(name):
    if pd.isna(name):
        return ""
    name = str(name)
    # Remove numbers
    name = re.sub(r'\d+', ' ', name)
    # Remove extra spaces
    name = re.sub(r'\s+', ' ', name).strip()
    return name

def get_name_no_numbers(name):
    name = normalize_name(name)

    # Remove numbers
    name = re.sub(r'\d+', ' ', name)

    # Remove extra spaces
    name = re.sub(r'\s+', ' ', name).strip()

    return name
```

```python
sample_pairs["s1_no_numbers"] = sample_pairs["s1_name"].apply(get_name_no_numbers)

sample_pairs["matched_no_numbers"] = sample_pairs["matched_name"].apply(get_name_no_numbers)
```

```python
def extract_domain_name(name):

    if pd.isna(name):
        return ""

    name = str(name).lower().strip()

    # Remove protocol
    name = re.sub(r'^https?://', '', name)

    # Remove www.
    name = re.sub(r'^www\.', '', name)

    # Remove path
    name = name.split('/')[0]

    # Check whether it looks like a domain
    if '.' in name:
        parts = name.split('.')

        # Remove common domain extensions
        if parts[-1] in ["com", "org", "net", "in", "co", "biz"]:
            return parts[0]

    return ""
```

```python
def extract_dba_name(name):

    if pd.isna(name):
        return ""

    name = normalize_name(name)

    match = re.search(r'\bdba\b\s+(.*)', name)

    if match:
        return match.group(1).strip()

    return ""
```

```python
sample_pairs["s1_dba"] = sample_pairs["s1_name"].apply(extract_dba_name)

sample_pairs["matched_dba"] = sample_pairs["matched_name"].apply(extract_dba_name)
```

```python
## Tf-idf vectorization
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity

vectorizer = TfidfVectorizer(
    analyzer="char",
    ngram_range=(2, 4)
)

all_names = pd.concat([
    sample_pairs["s1_norm"],
    sample_pairs["matched_norm"]
]).fillna("")

vectorizer.fit(all_names)

s1_vec = vectorizer.transform(sample_pairs["s1_norm"])
matched_vec = vectorizer.transform(sample_pairs["matched_norm"])

sample_pairs["name_tfidf_similarity"] = [
    cosine_similarity(s1_vec[i], matched_vec[i])[0][0]
    for i in range(len(sample_pairs))
]
```

```python
core_vectorizer = TfidfVectorizer(
    analyzer="char",
    ngram_range=(2, 4)
)

all_core_names = pd.concat([
    sample_pairs["s1_core"],
    sample_pairs["matched_core"]
]).fillna("")

core_vectorizer.fit(all_core_names)

s1_core_vec = core_vectorizer.transform(
    sample_pairs["s1_core"]
)

matched_core_vec = core_vectorizer.transform(
    sample_pairs["matched_core"]
)

sample_pairs["core_tfidf_similarity"] = [
    cosine_similarity(
        s1_core_vec[i],
        matched_core_vec[i]
    )[0][0]
    for i in range(len(sample_pairs))
]
```
