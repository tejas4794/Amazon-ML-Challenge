```python
## robust normalization(robust normalized name + core name)
import re
import unicodedata

def normalize_name(name):
    if pd.isna(name):
        return ""

    name = str(name).lower()

    # Normalize accented characters
    name = unicodedata.normalize("NFKD", name)
    name = "".join(
        ch for ch in name
        if not unicodedata.combining(ch)
    )

    # Remove URLs
    name = re.sub(r"https?://\S+|www\.\S+", " ", name)

    # Replace punctuation with spaces
    name = re.sub(r"[^a-z0-9\s]", " ", name)

    # Remove extra spaces
    name = re.sub(r"\s+", " ", name).strip()

    return name
```

```python
## extracting suffixes
SUFFIXES = [
    "private limited",
    "private ltd",
    "pvt limited",
    "pvt ltd",
    "limited",
    "ltd",
    "corporation",
    "corp",
    "incorporated",
    "inc",
    "company",
    "co",
    "llc",
    "llp"
]


def extract_suffix(name):

    name = normalize_name(name)

    for suffix in sorted(SUFFIXES, key=len, reverse=True):

        if name.endswith(" " + suffix):
            return suffix

    return ""
```

```python
## create the core names
def extract_core_name(name):

    name = normalize_name(name)

    suffix = extract_suffix(name)

    if suffix:
        name = name[:-len(suffix)].strip()

    return name
```

```python
## get better names

def remove_numbers(name):
    if pd.isna(name):
        return ""
    name = str(name)
    # Remove numbers
    name = re.sub(r'\d+', ' ', name)
    # Remove extra spaces
    name = re.sub(r'\s+', ' ', name).strip()
    return name

def get_name_no_numbers(name):
    name = normalize_name(name)

    # Remove numbers
    name = re.sub(r'\d+', ' ', name)

    # Remove extra spaces
    name = re.sub(r'\s+', ' ', name).strip()

    return name
```

```python
def extract_domain_name(name):

    if pd.isna(name):
        return ""

    name = str(name).lower().strip()

    # Remove protocol
    name = re.sub(r'^https?://', '', name)

    # Remove www.
    name = re.sub(r'^www\.', '', name)

    # Remove path
    name = name.split('/')[0]

    # Check whether it looks like a domain
    if '.' in name:
        parts = name.split('.')

        # Remove common domain extensions
        if parts[-1] in ["com", "org", "net", "in", "co", "biz"]:
            return parts[0]

    return ""
```

```python
def extract_dba_name(name):

    if pd.isna(name):
        return ""

    name = normalize_name(name)

    match = re.search(r'\bdba\b\s+(.*)', name)

    if match:
        return match.group(1).strip()

    return ""
```

```python
def token_overlap(a, b):
    a_tokens = set(normalize_name(a).split())
    b_tokens = set(normalize_name(b).split())

    if not a_tokens or not b_tokens:
        return 0.0

    intersection = len(a_tokens & b_tokens)
    union = len(a_tokens | b_tokens)

    return intersection / union
```

```python
def extract_numbers(text):
    if pd.isna(text):
        return set()

    return set(re.findall(r'\d+', str(text)))
```

```python
def number_overlap(a, b):
    a_nums = extract_numbers(a)
    b_nums = extract_numbers(b)

    if not a_nums or not b_nums:
        return 0.0

    return len(a_nums & b_nums) / len(a_nums | b_nums)
```

```python
## Candidate pool Genertion
hard_candidate_rows = []

rng = np.random.default_rng(42)

SAMPLE_PER_S1 = 500

for _, s1_row in s1_sample.iterrows():

    s1_id = s1_row["entity_id"]
    country = s1_row["country"]

    true_ids = true_matches.get(s1_id, set())

    # Pick either S2 or S3
    s2_group = s2_by_country.get(country)
    s3_group = s3_by_country.get(country)

    groups = []

    if s2_group is not None and len(s2_group) > 0:
        groups.append(s2_group)

    if s3_group is not None and len(s3_group) > 0:
        groups.append(s3_group)

    if not groups:
        continue

    # Randomly choose one source
    group = groups[rng.integers(len(groups))]

    # Sample candidates
    sample_size = min(SAMPLE_PER_S1, len(group))

    indices = rng.choice(
        len(group),
        size=sample_size,
        replace=False
    )

    sampled = group.iloc[indices]

    # Remove actual matches for THIS S1
    sampled = sampled[
        ~sampled["entity_id"].isin(true_ids)
    ]

    for _, candidate in sampled.iterrows():

        hard_candidate_rows.append({
            "s1_id": s1_id,
            "s1_name": s1_row["business_name"],
            "s1_address": s1_row["business_address"],
            "s1_country": country,

            "matched_id": candidate["entity_id"],
            "matched_name": candidate["business_name"],
            "matched_address": candidate["business_address"],
            "matched_country": candidate["country"]
        })

hard_candidates = pd.DataFrame(hard_candidate_rows)

print("Hard candidate pairs:", len(hard_candidates))
```

```python
## Calculate name TF-IDF similarity
hard_candidates["s1_norm"] = (
    hard_candidates["s1_name"]
    .apply(normalize_name)
)

hard_candidates["matched_norm"] = (
    hard_candidates["matched_name"]
    .apply(normalize_name)
)
```

```python
hc_s1_vec = vectorizer.transform(
    hard_candidates["s1_norm"]
)

hc_matched_vec = vectorizer.transform(
    hard_candidates["matched_norm"]
)
```

```python
hard_candidates["name_similarity"] = (
    hc_s1_vec.multiply(hc_matched_vec)
    .sum(axis=1)
    .A1
)
```

```python
## Step 3 — Pick the hardest negative for each S1
hard_negatives = (
    hard_candidates
    .sort_values(
        ["s1_id", "name_similarity"],
        ascending=[True, False]
    )
    .groupby("s1_id")
    .head(1)
    .copy()
)
```

```python
# Step 4 — Make sure they really are negatives
hard_negatives["is_true_match"] = hard_negatives.apply(
    lambda row:
        row["matched_id"] in true_matches.get(row["s1_id"], set()),
    axis=1
)

print(
    hard_negatives["is_true_match"].value_counts()
)
```

```python
# Step 5 — Inspect our hard negatives
display(
    hard_negatives[[
        "s1_name",
        "matched_name",
        "name_similarity",
        "s1_address",
        "matched_address"
    ]]
    .sort_values("name_similarity", ascending=False)
    .head(30)
)
```

```python
# Prepare hard negatives
hard_negatives["s1_norm"] = hard_negatives["s1_name"].apply(normalize_name)
hard_negatives["matched_norm"] = hard_negatives["matched_name"].apply(normalize_name)

hard_negatives["s1_core"] = hard_negatives["s1_name"].apply(extract_core_name)
hard_negatives["matched_core"] = hard_negatives["matched_name"].apply(extract_core_name)

hard_negatives["s1_address_norm"] = hard_negatives["s1_address"].apply(normalize_address)
hard_negatives["matched_address_norm"] = hard_negatives["matched_address"].apply(normalize_address)
```

```python
# TF-IDF similarities
s1_vec_h = vectorizer.transform(hard_negatives["s1_norm"].fillna(""))
matched_vec_h = vectorizer.transform(hard_negatives["matched_norm"].fillna(""))

s1_core_vec_h = core_vectorizer.transform(
    hard_negatives["s1_core"].fillna("")
)

matched_core_vec_h = core_vectorizer.transform(
    hard_negatives["matched_core"].fillna("")
)

s1_addr_vec_h = address_vectorizer.transform(
    hard_negatives["s1_address_norm"].fillna("")
)

matched_addr_vec_h = address_vectorizer.transform(
    hard_negatives["matched_address_norm"].fillna("")
)
```

```python
hard_negatives["name_tfidf_similarity"] = (
    s1_vec_h.multiply(matched_vec_h).sum(axis=1).A1
)

hard_negatives["core_tfidf_similarity"] = (
    s1_core_vec_h.multiply(matched_core_vec_h).sum(axis=1).A1
)

hard_negatives["address_tfidf_similarity"] = (
    s1_addr_vec_h.multiply(matched_addr_vec_h).sum(axis=1).A1
)
```

```python
# Token overlap
hard_negatives["name_token_overlap"] = hard_negatives.apply(
    lambda row: token_overlap(
        row["s1_name"],
        row["matched_name"]
    ),
    axis=1
)
```

```python
# Number features
hard_negatives["name_number_overlap"] = hard_negatives.apply(
    lambda row: number_overlap(
        row["s1_name"],
        row["matched_name"]
    ),
    axis=1
)

hard_negatives["address_number_overlap"] = hard_negatives.apply(
    lambda row: number_overlap(
        row["s1_address"],
        row["matched_address"]
    ),
    axis=1
)
```

```python
# Country match
hard_negatives["country_match"] = (
    hard_negatives["s1_country"] ==
    hard_negatives["matched_country"]
).astype(int)
```

```python
# Label them
hard_negatives["label"] = 0
```

```python
X_train = train_data[features]
y_train = train_data["label"]

X_val = val_data[features]
y_val = val_data["label"]
```

```python
print(training_data.shape)
print(training_data["label"].value_counts())

print(train_data.shape)
print(val_data.shape)
```

```python
from sklearn.linear_model import LogisticRegression

model = LogisticRegression(
    max_iter=1000,
    class_weight="balanced",
    random_state=42
)

model.fit(X_train, y_train)
```

```python
val_probs = model.predict_proba(X_val)[:, 1]
```

```python
from sklearn.metrics import classification_report

val_pred = (val_probs >= 0.5).astype(int)

print(classification_report(y_val, val_pred))
```

```python
from sklearn.metrics import fbeta_score

f05 = fbeta_score(
    y_val,
    val_pred,
    beta=0.5
)

print("F0.5:", f05)
```

```python
from sklearn.metrics import precision_score, recall_score
```

```python
import numpy as np

thresholds = np.arange(0.1, 0.96, 0.01)

results = []

for threshold in thresholds:
    pred = (val_probs >= threshold).astype(int)

    precision = precision_score(y_val, pred, zero_division=0)
    recall = recall_score(y_val, pred, zero_division=0)
    f05 = fbeta_score(
        y_val,
        pred,
        beta=0.5,
        zero_division=0
    )

    results.append(
        (threshold, precision, recall, f05)
    )
```

```python
results_df = pd.DataFrame(
    results,
    columns=["threshold", "precision", "recall", "f05"]
)

display(
    results_df.sort_values(
        "f05",
        ascending=False
    ).head(10)
)
```

```python
best_threshold = 0.39

val_results = val_data.copy()
val_results["probability"] = val_probs
val_results["prediction"] = (
    val_probs >= best_threshold
).astype(int)

false_positives = val_results[
    (val_results["label"] == 0) &
    (val_results["prediction"] == 1)
]

false_negatives = val_results[
    (val_results["label"] == 1) &
    (val_results["prediction"] == 0)
]

print("False positives:", len(false_positives))
print("False negatives:", len(false_negatives))
```
